#1. What is the national trend in the average "Percent of adults aged 18 years and older who have obesity" from 2011 to 2023?

```python
# 1. Filter the data for the specific question on obesity
obese_question = 'Percent of adults aged 18 years and older who have obesity'
obese_data = healthdata[
    (healthdata['Class'] == 'Obesity / Weight Status') &
    (healthdata['Question'] == obese_question)
].copy()

# 2. Group by year and calculate the national average obesity rate
obesity_trend = obese_data.groupby('YearStart')['Data_Value'].mean().reset_index()
obesity_trend.columns = ['Year', 'Average_Obesity_Rate']

# 3. Create the line plot
plt.figure(figsize=(7, 4))
sns.lineplot(
    data=obesity_trend,
    x='Year',
    y='Average_Obesity_Rate',
    marker='o',
    color='darkorange',
    linewidth=2
)
plt.title('Average US Adult Obesity Rate Trend (2011-2023)', fontsize=10)
plt.xlabel('Year', fontsize=10)
plt.ylabel('Average Obesity Rate (%)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
```

# INSIGHTS
Based on the line plot, the national average obesity rate for U.S. adults has shown a clear and consistent increasing trend from 2011 to 2023. Starting around an average of $27.7\%$ in 2011, the rate climbed steadily each year until 2019, where it reached approximately $32.3\%$. After a slight dip or plateau around 2020, there was a noticeable jump in 2021, and the rate appears to be leveling off at a new, higher average of about $34.0\%$ in the most recent years (2021-2023). This visual suggests that, despite public health efforts, the average prevalence of adult obesity continues to be a growing concern over the last decade.

# CONCLUSION
The analysis of the Behavioral Risk Factor Surveillance System data clearly demonstrates a persistent upward trend in the average adult obesity rate from 2011 to 2023. This continuous rise indicates a significant public health challenge that requires ongoing attention from policymakers and healthcare professionals.

 # 2. Which age group has the highest average percentage of adults who report engaging in no leisure-time physical activity across the United States in the latest year of the data (2023)?

```python

# Define the filtering criteria
target_year = healthdata['YearStart'].max()
target_question = 'Percent of adults who engage in no leisure-time physical activity'
target_class = 'Physical Activity'
# Corrected stratification category name to include a space
target_stratification_category = 'Age (years)'

# 1. Filter the data
physical_activity_by_age = healthdata[
    (healthdata['YearStart'] == target_year) &
    (healthdata['Class'] == target_class) &
    (healthdata['Question'] == target_question) &
    (healthdata['StratificationCategory1'] == target_stratification_category)
].copy()

# 2. Aggregate the data: Calculate the average rate for each age group
activity_rate_by_age = physical_activity_by_age.groupby('Stratification1')['Data_Value'].mean().reset_index()
activity_rate_by_age.columns = ['Age_Group', 'Average_No_Leisure_Activity_Rate']

# 3. Sort the data by rate in descending order for the bar plot
activity_rate_by_age_sorted = activity_rate_by_age.sort_values(
    'Average_No_Leisure_Activity_Rate', ascending=False
)

# 4. Create the bar plot (matrix)
plt.figure(figsize=(7, 4))
sns.barplot(
    data=activity_rate_by_age_sorted,
    x='Age_Group',
    y='Average_No_Leisure_Activity_Rate',
    palette='magma'
)

plt.title(f'Average No Leisure-Time Physical Activity Rate by Age Group ({target_year})', fontsize=10)
plt.xlabel('Age Group (Years)', fontsize=10)
plt.ylabel('Average Rate (%)', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

# INSIGHTS
The bar plot clearly shows that the average percentage of adults who do not engage in any leisure-time physical activity is strongly linked to age. The highest rate of inactivity, at about $30.8\%$, belongs to the $65$ years or older age group, indicating older adults are the least active. Following closely behind is the $55-64$ years group, with an average rate of approximately $27.5\%$. In contrast, the younger age groups, specifically $18-24$ years and $25-34$ years, show the lowest rates of inactivity, suggesting younger adults are the most likely to participate in some form of leisure-time physical activity.

# CONCLUSION
The data establishes a clear pattern: as age increases, the proportion of adults who report having no leisure-time physical activity also increases, reaching its peak in the senior population. This trend is consistent with known health behavior patterns, where mobility issues, chronic illness, and other factors often lead to reduced physical activity later in life. This finding is crucial for public health programs, suggesting that interventions aimed at promoting physical activity should specifically target older adults with accessible, low-impact exercise options to maintain health and quality of life.

# 3. What is the relationship between a location's average adult obesity rate and its average rate of no leisure-time physical activity across all available years in the dataset?

```python


# Define the two questions and classes
obesity_q = 'Percent of adults aged 18 years and older who have obesity'
activity_q = 'Percent of adults who engage in no leisure-time physical activity'
obesity_class = 'Obesity / Weight Status'
activity_class = 'Physical Activity'

# 1. Prepare Obesity Data: Aggregate by LocationDesc and YearStart
obesity_data = healthdata[
    (healthdata['Class'] == obesity_class) &
    (healthdata['Question'] == obesity_q)
].copy()
obesity_agg = obesity_data.groupby(['LocationDesc', 'YearStart'])['Data_Value'].mean().reset_index()
obesity_agg.rename(columns={'Data_Value': 'Avg_Obesity_Rate'}, inplace=True)

# 2. Prepare Activity Data: Aggregate by LocationDesc and YearStart
activity_data = healthdata[
    (healthdata['Class'] == activity_class) &
    (healthdata['Question'] == activity_q)
].copy()
activity_agg = activity_data.groupby(['LocationDesc', 'YearStart'])['Data_Value'].mean().reset_index()
activity_agg.rename(columns={'Data_Value': 'Avg_Inactivity_Rate'}, inplace=True)

# 3. Merge Data: Join the two dataframes on Location and Year
merged_data = pd.merge(obesity_agg, activity_agg, on=['LocationDesc', 'YearStart'])

# 4. Create Scatterplot (matrix)
plt.figure(figsize=(7, 4))
sns.scatterplot(
    data=merged_data,
    x='Avg_Inactivity_Rate',
    y='Avg_Obesity_Rate',
    color='purple',
    alpha=0.6,
    marker='o'
)

# Apply simple formatting
plt.title('Relationship Between Obesity and Physical Inactivity (2011-2023)', fontsize=14)
plt.xlabel('Average No Leisure-Time Physical Activity Rate (%)', fontsize=12)
plt.ylabel('Average Obesity Rate (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.gcf().set_size_inches(7, 4)
```

# INSIGHTS
The scatterplot shows a clear positive correlation between the average rate of physical inactivity and the average rate of obesity. This means that data points representing locations and years with high rates of no leisure-time physical activity also tend to have high obesity rates. Conversely, the locations and years with the lowest obesity rates are clustered where the inactivity rates are also low. The data forms a recognizable upward-sloping cloud, strongly suggesting that lack of exercise and weight gain go hand-in-hand across different populations.

# CONCLUSION
The strong positive linear relationship between physical inactivity and obesity rates confirms that physical activity is a major public health factor impacting weight status. States and years with less active populations show consistently higher levels of obesity. This matrix validates the importance of promoting regular leisure-time physical activity as a primary strategy for tackling the nation's obesity epidemic. Therefore, policy focus should continue to support accessible and encouraging environments for people to move more and reduce sedentary time.

Centers for Disease Control and Prevention. Physical Activity and Health. Link: https://www.cdc.gov/physicalactivity/about-physical-activity/index.htm

4. How has the average adult obesity rate (percentage of adults aged $18$ years and older who have obesity) for males compared to females changed over the years from 2011 to 2023?

```python

# 1. Define the filtering criteria
target_question = 'Percent of adults aged 18 years and older who have obesity'
target_class = 'Obesity / Weight Status'
target_stratification_category = 'Sex'

# 2. Filter the data for the specific question and stratification
obesity_by_sex_data = healthdata[
    (healthdata['Class'] == target_class) &
    (healthdata['Question'] == target_question) &
    (healthdata['StratificationCategory1'] == target_stratification_category)
].copy()

# 3. Aggregate the data: Calculate the average obesity rate by Year and Sex
obesity_trend_by_sex = obesity_by_sex_data.groupby(['YearStart', 'Stratification1'])['Data_Value'].mean().reset_index()
obesity_trend_by_sex.columns = ['Year', 'Sex', 'Average_Obesity_Rate']

# 4. Create the line plot (matrix)
plt.figure(figsize=(7, 4))
# Using lineplot with hue to plot two lines (Male and Female)
sns.lineplot(
    data=obesity_trend_by_sex,
    x='Year',
    y='Average_Obesity_Rate',
    hue='Sex',
    marker='o',
    linestyle='solid'
)

# Apply simple formatting
plt.title('Average Adult Obesity Rate Trend by Sex (2011-2023)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Obesity Rate (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(obesity_trend_by_sex['Year'].unique(), rotation=45)
plt.legend(title='Sex')
plt.tight_layout()
```

#INSIGHTS
Looking at the line plot, both males and females showed a consistent upward trend in their average obesity rates from 2011 to 2023, meaning obesity has increased for everyone. The data shows that males generally had a slightly higher average rate than females between 2011 and 2018. However, a noticeable shift happened around 2018-2019, after which the average obesity rate for females surpassed that of males and continued to rise more steeply. By 2023, the average obesity rate for females was approximately  34.5% , which is higher than the rate of  32.8%  observed for males, suggesting that in recent years, females are experiencing a faster increase in average obesity prevalence.

# CONCLUSION
The analysis comparing obesity trends by sex reveals that while the obesity rate has climbed steadily for both groups over the last decade, the pattern has shifted. Females have shown a greater rate of increase in average obesity since 2018, leading to a noticeable gap where their average rate is now higher than the male average. This suggests that public health efforts aimed at curbing obesity may need to be tailored to better address the specific health and lifestyle factors contributing to this accelerating trend among women. Targeted research into the causes of this recent divergence is warranted to develop effective intervention strategies.

# 5. What is the relationship between a location's average rate of low daily fruit consumption (less than one time daily) and its average rate of no leisure-time physical activity?

```python


# Define the two questions and classes
fruit_q = 'Percent of adults who report consuming fruit less than one time daily'
activity_q = 'Percent of adults who engage in no leisure-time physical activity'
nutrition_class = 'Fruits and Vegetables'
activity_class = 'Physical Activity'

# 1. Prepare Low Fruit Consumption Data: Aggregate by LocationDesc and YearStart
fruit_data = healthdata[
    (healthdata['Class'] == nutrition_class) &
    (healthdata['Question'] == fruit_q)
].copy()
fruit_agg = fruit_data.groupby(['LocationDesc', 'YearStart'])['Data_Value'].mean().reset_index()
fruit_agg.rename(columns={'Data_Value': 'Avg_Low_Fruit_Rate'}, inplace=True)

# 2. Prepare Activity Data: Aggregate by LocationDesc and YearStart
activity_data = healthdata[
    (healthdata['Class'] == activity_class) &
    (healthdata['Question'] == activity_q)
].copy()
activity_agg = activity_data.groupby(['LocationDesc', 'YearStart'])['Data_Value'].mean().reset_index()
activity_agg.rename(columns={'Data_Value': 'Avg_Inactivity_Rate'}, inplace=True)

# 3. Merge Data: Join the two dataframes on Location and Year
merged_data = pd.merge(fruit_agg, activity_agg, on=['LocationDesc', 'YearStart'])

# 4. Create Scatterplot (matrix)
plt.figure(figsize=(7, 4))
sns.scatterplot(
    data=merged_data,
    x='Avg_Inactivity_Rate',
    y='Avg_Low_Fruit_Rate',
    color='darkorange',
    alpha=0.7,
    marker='D'
)

# Apply simple formatting
plt.title('Relationship Between Low Fruit Consumption and Physical Inactivity', fontsize=10)
plt.xlabel('Average No Leisure-Time Physical Activity Rate (%)', fontsize=10)
plt.ylabel('Average Low Fruit Consumption Rate (%)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.gcf().set_size_inches(7, 4)

```

# INSIGHTS
The scatterplot shows a clear positive correlation between the two variables, although the relationship is not extremely tight. Data points where the inactivity rate is high also tend to have a high rate of low fruit consumption, suggesting these poor habits often occur together in a population. This visually suggests that locations that are physically active are also more likely to have better dietary habits.

# CONCLUSION
The strong positive correlation between physical inactivity and low fruit consumption confirms that health behaviors are often linked at the population level. Locations that struggle with one negative health behavior, such as lack of exercise, often struggle with another, like poor diet. This finding is important for designing effective health campaigns that encourage multiple positive changes at once. Targeting both diet and activity simultaneously could lead to more significant improvements in public health outcomes.

# 8. Which are the top 10 states or territories that report the highest average adult obesity rate based on all available data in the dataset?

```python

# 1. Filter data for Adult Obesity Question
obesity_data = healthdata[
    (healthdata['Class'] == 'Obesity / Weight Status') &
    (healthdata['Question'] == 'Percent of adults aged 18 years and older who have obesity')
].copy()

# Drop records where the rate is missing
obesity_data.dropna(subset=['Data_Value'], inplace=True)

# 2. Aggregate: Calculate the overall average obesity rate for each location
avg_obesity_by_location = obesity_data.groupby('LocationDesc')['Data_Value'].mean().reset_index()
avg_obesity_by_location.columns = ['Location', 'Average_Obesity_Rate']

# 3. Parameter: Find the top 10 locations with the highest rate
top_10_obesity = avg_obesity_by_location.nlargest(10, 'Average_Obesity_Rate')

# 4. Create the Bar Plot (matrix)
plt.figure(figsize=(7,4))
sns.barplot(
    data=top_10_obesity,
    x='Location',
    y='Average_Obesity_Rate',
    palette='Reds_d'
)
plt.title('Top 10 Locations with the Highest Average Adult Obesity Rate', fontsize=10)
plt.xlabel('State/Territory', fontsize=10)
plt.ylabel('Average Obesity Rate (%)', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.gcf().set_size_inches(7,4)

```

# INSIGHTS
The bar plot reveals that the top 10 locations with the highest average obesity rates are all US states, with West Virginia having the highest rate at approximately $37.8\%$. This visual ranking quickly identifies the areas where the obesity epidemic is most concentrated and severe.Conclusion

# CONCLUSION
The finding that West Virginia and Mississippi lead the nation in average obesity rates highlights the need for targeted health funding and resources in these areas. The strong regional concentration suggests that shared factors, such as economic conditions, cultural norms, and access to healthy food, may be at play. # 10. How does the statistical distribution (median, spread, and outliers) of the reported adult obesity rate vary across different education levels?

```python

# 1. Filter data for Adult Obesity Question and Education stratification
target_question = 'Percent of adults aged 18 years and older who have obesity'
target_class = 'Obesity / Weight Status'
target_stratification_category = 'Education'

obesity_education_data = healthdata[
    (healthdata['Class'] == target_class) &
    (healthdata['Question'] == target_question) &
    (healthdata['StratificationCategory1'] == target_stratification_category)
].copy()

# Ensure Data_Value is numeric and drop missing values
obesity_education_data['Data_Value'] = pd.to_numeric(obesity_education_data['Data_Value'], errors='coerce')
obesity_education_data.dropna(subset=['Data_Value'], inplace=True)

# 2. Define the order for the boxplot categories (Education)
education_order = [
    'Less than high school',
    'High school graduate',
    'Some college or technical school',
    'College graduate',
    'Data not reported'
]
available_education_levels = obesity_education_data['Stratification1'].unique()
ordered_levels = [level for level in education_order if level in available_education_levels]

# 3. Create the Box Plot (matrix)
plt.figure(figsize=(7, 4))
sns.boxplot(
    data=obesity_education_data,
    x='Stratification1',
    y='Data_Value',
    order=ordered_levels,
    palette='mako'
)

# Apply simple formatting
plt.title('Distribution of Adult Obesity Rates (%) by Education Level', fontsize=10)
plt.xlabel('Education Level', fontsize=10)
plt.ylabel('Obesity Rate (%)', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.gcf().set_size_inches(7,4)

```

# INSIGHTS
The boxplot clearly demonstrates that as educational attainment increases, the median obesity rate consistently decreases. The 'Less than high school' group has the highest median obesity rate, and the 'College graduate' group has the lowest, showing a strong inverse relationship. Furthermore, the boxes for lower education levels are wider and generally higher on the y-axis, indicating a greater range of high obesity rates across different locations. This suggests that higher education is generally associated with a reduced risk of obesity.

# CONCLUSION
The strong visual trend confirms that education level is a significant factor linked to adult obesity rates, reflecting a clear socioeconomic gradient in health outcomes. The consistent drop in the median rate with each step up in education suggests that knowledge, access to resources, and economic opportunity play a key role in maintaining a healthy weight. Public health initiatives aimed at reducing obesity should therefore include education-based programs. Policies that promote higher educational attainment may also contribute indirectly to better overall population health.

# 11. How have the average adult obesity rates in the five locations with the highest overall prevalence changed year-over-year from 2020 to the latest available year (2023)?

```python

# Define the filtering criteria
target_question = 'Percent of adults aged 18 years and older who have obesity'
target_class = 'Obesity / Weight Status'
start_year = 2020
latest_year = healthdata['YearStart'].max()

# 1. Filter data for the obesity question
obesity_data = healthdata[
    (healthdata['Class'] == target_class) &
    (healthdata['Question'] == target_question)
].copy()
obesity_data['Data_Value'] = pd.to_numeric(obesity_data['Data_Value'], errors='coerce')
obesity_data.dropna(subset=['Data_Value'], inplace=True)

# 2. Filter data for year >= 2020
recent_obesity_data = obesity_data[obesity_data['YearStart'] >= start_year].copy()

# 3. Identify 5 locations with the highest *overall* average obesity rate
avg_obesity_by_location = recent_obesity_data.groupby('LocationDesc')['Data_Value'].mean().reset_index()
top_5_locations = avg_obesity_by_location.nlargest(5, 'Data_Value')['LocationDesc'].tolist()

# 4. Filter the recent data to include ONLY the top 5 locations
final_heatmap_data = recent_obesity_data[
    recent_obesity_data['LocationDesc'].isin(top_5_locations)
].copy()

# 5. Aggregate and Pivot the data for the heatmap
obesity_rate_by_year_location = final_heatmap_data.groupby(['YearStart', 'LocationDesc'])['Data_Value'].mean().reset_index()

heatmap_pivot = obesity_rate_by_year_location.pivot(
    index='YearStart',
    columns='LocationDesc',
    values='Data_Value'
)
heatmap_pivot = heatmap_pivot[top_5_locations] # Keep the top 5 order

# 6. Create the Heatmap (matrix)
plt.figure(figsize=(7, 4))
sns.heatmap(
    heatmap_pivot,
    annot=True,
    fmt=".1f",
    linewidths=.5,
    cmap='Reds',
    cbar_kws={'label': 'Average Adult Obesity Rate (%)'}
)

# Apply simple formatting
plt.title(f'Avg Adult Obesity Rate (%) in Top 5 Locations ({start_year}-{latest_year})', fontsize=10)
plt.ylabel('Year', fontsize=10)
plt.xlabel('Location', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.gcf().set_size_inches(7, 4)


```

# INSIGHTS
The heatmap shows that all five of these high-prevalence locations maintained very high average obesity rates, consistently hovering around $38\%$ to $41\%$ between 2020 and 2023. The data also suggests that the rates generally increased from 2020 to 2021 across most of these locations, possibly reflecting changes in lifestyle during the initial years of the COVID-19 pandemic. However, there is no clear trend of significant improvement or reduction in the high obesity levels for any of the top 5 locations over this period.
